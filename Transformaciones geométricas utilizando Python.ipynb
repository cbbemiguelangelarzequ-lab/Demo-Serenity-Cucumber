```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

# 1. Cargamos la imamgen
imagen = cv2.imread('imagen.jpg')

# Verificamos que la imagen se subió correctamente
if imagen is None:
    print("Error: No se encontró la imagen. Asegúrate de haberla subido a la pestaña de Archivos en Colab.")
else:
    altura, anchura = imagen.shape[:2]
    print("--- IMAGEN ORIGINAL ---")
    cv2_imshow(imagen) # Mostramos la imagen original

    # A. TRASLACIÓN
    # Movemos la imagen 100 píxeles a la derecha y 50 hacia abajo
    tx, ty = 100, 50
    M_traslacion = np.float32([[1, 0, tx], [0, 1, ty]])
    img_trasladada = cv2.warpAffine(imagen, M_traslacion, (anchura, altura))

    print("\n--- RESULTADO: TRASLACIÓN ---")
    cv2_imshow(img_trasladada)

    # B. ROTACIÓN
    # Rotamos 45 grados desde el centro
    centro = (anchura // 2, altura // 2)
    angulo = 45
    escala = 1.0
    M_rotacion = cv2.getRotationMatrix2D(centro, angulo, escala)
    img_rotada = cv2.warpAffine(imagen, M_rotacion, (anchura, altura))

    print("\n--- RESULTADO: ROTACIÓN ---")
    cv2_imshow(img_rotada)

    # C. ESCALADO
    # Reducimos la imagen a la mitad de su tamaño (0.5)
    img_escalada = cv2.resize(imagen, None, fx=0.5, fy=0.5, interpolation=cv2.INTER_LINEAR)

    print("\n--- RESULTADO: ESCALADO ---")
    cv2_imshow(img_escalada)
```
